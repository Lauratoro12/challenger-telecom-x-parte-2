```python
from sklearn.utils import resample

# Dividir en clases mayoritaria y minoritaria
df_majority = df[df['Churn'] == 'No']
df_minority = df[df['Churn'] == 'Yes']

# Submuestrear la clase mayoritaria
df_majority_downsampled = resample(df_majority,
                                   replace=False,    # No reemplazar
                                   n_samples=len(df_minority),  # Mismo tamaño que la clase minoritaria
                                   random_state=42)  # Para reproducibilidad

# Combinar clases minoritaria y submuestreada
df_downsampled = pd.concat([df_majority_downsampled, df_minority])

# Ver la nueva distribución
print(df_downsampled['Churn'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Drop rows where 'Churn' is NaN
df.dropna(subset=['Churn'], inplace=True)

# Convert 'Churn' column to numerical (0 for No, 1 for Yes)
df['Churn'] = df['Churn'].map({'No': 0, 'Yes': 1})

# Identify all remaining categorical columns (object type) for one-hot encoding
categorical_cols_to_encode = df.select_dtypes(include='object').columns

# Apply One-Hot Encoding to the remaining categorical columns
df = pd.get_dummies(df, columns=categorical_cols_to_encode, drop_first=True)

# Variables independientes (features)
X = df.drop('Churn', axis=1)

# Variable dependiente (target)
y = df['Churn']
```

```python
# División de los datos
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,   # 20% para prueba
    random_state=42, # Para reproducibilidad
    stratify=y       # Mantener la proporción de clases Churn
)

# Verificar tamaños
print("Tamaño de X_train:", X_train.shape)
print("Tamaño de X_test:", X_test.shape)
print("Tamaño de y_train:", y_train.shape)
print("Tamaño de y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Seleccionar columnas numéricas
numerical_cols = ['tenure', 'Monthly', 'Total', 'Cuentas_Diarias']

# Inicializar el scaler
scaler = StandardScaler()

# Aplicar la estandarización solo a las columnas numéricas
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test_scaled[numerical_cols] = scaler.transform(X_test[numerical_cols])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Inicializar el modelo
log_model = LogisticRegression(random_state=42, max_iter=1000)

# Entrenar el modelo con los datos normalizados
log_model.fit(X_train_scaled, y_train)

# Predecir en el conjunto de prueba
y_pred_log = log_model.predict(X_test_scaled)

# Evaluar el modelo
print("=== Regresión Logística ===")
print("Accuracy:", accuracy_score(y_test, y_pred_log))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_log))
print("\nClassification Report:\n", classification_report(y_test, y_pred_log))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Inicializar el modelo
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# Entrenar el modelo sin normalización
rf_model.fit(X_train, y_train)

# Predecir en el conjunto de prueba
y_pred_rf = rf_model.predict(X_test)

# Evaluar el modelo
print("=== Random Forest ===")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Exactitud
accuracy_log = accuracy_score(y_test, y_pred_log)

# Precisión
precision_log = precision_score(y_test, y_pred_log)

# Recall
recall_log = recall_score(y_test, y_pred_log)

# F1-score
f1_log = f1_score(y_test, y_pred_log)

# Matriz de Confusión
conf_matrix_log = confusion_matrix(y_test, y_pred_log)

# Resultados
print("=== Métricas de Regresión Logística ===")
print("Exactitud:", accuracy_log)
print("Precisión:", precision_log)
print("Recall:", recall_log)
print("F1-score:", f1_log)
print("\nMatriz de Confusión:\n", conf_matrix_log)
```

```python
# Exactitud
accuracy_rf = accuracy_score(y_test, y_pred_rf)

# Precisión
precision_rf = precision_score(y_test, y_pred_rf)

# Recall
recall_rf = recall_score(y_test, y_pred_rf)

# F1-score
f1_rf = f1_score(y_test, y_pred_rf)

# Matriz de Confusión
conf_matrix_rf = confusion_matrix(y_test, y_pred_rf)

# Resultados
print("=== Métricas de Random Forest ===")
print("Exactitud:", accuracy_rf)
print("Precisión:", precision_rf)
print("Recall:", recall_rf)
print("F1-score:", f1_rf)
print("\nMatriz de Confusión:\n", conf_matrix_rf)
```

```python
# Usamos el modelo KNN para calcular la distancia
from sklearn.neighbors import KNeighborsClassifier
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled, y_train)

# Mostrar los coeficientes de importancia
# KNN no tiene coeficientes directos, pero podemos revisar la distancia y el peso de las variables
importances_knn = pd.DataFrame({
    'Variable': X_train.columns,
    'Importancia': knn_model.feature_importances_ if hasattr(knn_model, 'feature_importances_') else [0] * len(X_train.columns)
})

# Ordenar y ver las más importantes
importances_knn = importances_knn.sort_values(by='Importancia', ascending=False)
print(importances_knn.head())
```

Objetivo del Análisis

El propósito de este análisis es identificar los factores que influyen en la cancelación de clientes (Churn) y entrenar modelos predictivos para predecir la probabilidad de cancelación.

Modelos Utilizados

Regresión Logística: Modelo lineal que estima la probabilidad de que un cliente cancele.

Métricas:

Exactitud: 79%

Precisión: 73%

Recall: 45%

F1-score: 55%

Variables clave:

Tenure (tiempo de contrato) tiene un coeficiente negativo: más tiempo, menor probabilidad de cancelación.

MonthlyCharges (gasto mensual) tiene un coeficiente positivo: más gasto, mayor probabilidad de cancelación.

Random Forest: Modelo basado en árboles que maneja interacciones complejas entre variables.

Métricas:

Exactitud: 82%

Precisión: 80%

Recall: 70%

F1-score: 75%

Variables clave:

Tenure es la más importante. Los clientes con más tiempo en el servicio tienden a no cancelar.

MonthlyCharges y TotalCharges son relevantes, ya que los clientes con mayor gasto tienden a cancelar.

Factores Clave que Afectan la Cancelación

Tiempo de contrato (Tenure): Los clientes con menos tiempo en el servicio son más propensos a cancelar.

Gasto mensual (MonthlyCharges) y total (TotalCharges): Los clientes que gastan más tienen más probabilidad de cancelar, especialmente si no están satisfechos con el servicio.

Soporte técnico (TechSupport): Los clientes sin soporte técnico son más propensos a cancelar.

Recomendaciones de Estrategia

Ofrecer incentivos a clientes nuevos (menos de 6 meses) para reducir el churn temprano.

Revisar planes de clientes con alto gasto mensual y asegurarse de que reciban el valor esperado.

Mejorar el soporte técnico, especialmente para clientes sin acceso a esta opción.

Conclusión

Random Forest mostró el mejor rendimiento, especialmente al manejar interacciones entre variables.

Regresión Logística fue útil para obtener una visión clara de cómo cada variable afecta la cancelación, pero mostró underfitting.

Las variables más influyentes son el tiempo de contrato y el gasto mensual.
